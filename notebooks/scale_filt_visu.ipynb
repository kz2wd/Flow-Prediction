```python
import numpy as np
from space_exploration.beans.dataset_bean import Dataset
from space_exploration.dataset.transforms.AllTransforms import TransformationReferences
dataset_name = "re200-sr05etot"
dataset = Dataset.get_dataset_or_fail(dataset_name)
ds = dataset.get_training_dataset(64, TransformationReferences.SCALE_FILTERING.transformation, TransformationReferences.Y_ALONG_COMPONENT_NORMALIZE.transformation, size=100)
xs = np.array([x.numpy() for x, y in ds])
ys = np.array([y.numpy() for x, y in ds])

ds_normed = dataset.get_training_dataset(64, TransformationReferences.SCALE_FILTERING_RENORMALIZED.transformation, TransformationReferences.Y_ALONG_COMPONENT_NORMALIZE.transformation, size=100)
xs_normed = np.array([x.numpy() for x, y in ds_normed])
ys_normed = np.array([y.numpy() for x, y in ds_normed])
```

```python
import matplotlib.pyplot as plt

def plot_hist(i):
    names = [
        "Normalized Pressure U",
        "Normalized Shear Stress X",
        "Normalized Shear Stress Z",
        "Filter Pressure [9.172, 26.348]",
        "Filter Pressure [5.496, 29.237]",
        "Filter Pressure [10.792, 16.214]",
        "Filter Pressure [3.810, 23.111]",
        "Filter Pressure [5.496, 29.237]",
        "Filter Pressure [17.230, 26.774]",
        "Filter Pressure [27.775, 29.892]",
        "Filter Pressure [28.977, 35.500]",
        "Filter Stress X [5.496, 29.237]",
        "Filter Stress X [9.172, 26.348]",
        "Filter Stress X [28.977, 35.500]",
        "Filter Stress Z [3.810, 23.111]",
        "Filter Stress Z [5.496, 29.237]",
        "Filter Stress Z [9.172, 26.348]",
        "Filter Stress Z [10.792, 16.214]",
    ]
    fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(20, 5))
    ax1.hist(xs[:, i].flatten(), bins="auto")
    ax2.hist((xs[:, i] / xs[:, i].std()).flatten(), bins="auto")
    ax3.hist(xs_normed[:, i].flatten(), bins="auto")
    fig.suptitle(f"Histogram of {names[i]}")
    plt.show()
```
