```python
import ray
from ax.service.managed_loop import optimize

ray.init()
```

```python

@ray.remote
def run_optimization(i):
    
    best_parameters, values, experiment = optimize(
        parameters=[
            {"name": "x", "type": "range", "bounds": [0.0, 1.0]},
            {"name": "y", "type": "range", "bounds": [0.0, 1.0]},
        ],
        evaluation_function=lambda p: (p["x"] - 0.5) ** 2 + (p["y"] - 0.5) ** 2,
        objective_name="minimize",
        total_trials=20,
        random_seed=i,
    )
    return i
    print(best_parameters)
    return 0

seeds = [42, 43, 44]
futures = [run_optimization.remote(i) for i in range(3)]
ray.get(futures)

# print(results)
```
